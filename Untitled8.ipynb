```python
app_code = '''

import streamlit as st
import pandas as pd
import sqlite3
from datetime import datetime
from sklearn.linear_model import LinearRegression
import numpy as np
import plotly.express as px

# PAGE CONFIG


st.set_page_config(
    page_title="AI Food Ordering System",
    page_icon="🍔",
    layout="wide"
)

st.title("🍔 AI Smart Food Ordering System")


# DATABASE


conn = sqlite3.connect("food_app.db", check_same_thread=False)
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS orders (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    customer TEXT,
    item TEXT,
    quantity INTEGER,
    price INTEGER,
    order_time TEXT
)
""")

conn.commit()

# MENU


menu = {
    "Burger": 120,
    "Pizza": 250,
    "Pasta": 180,
    "French Fries": 100,
    "Coke": 50,
    "Ice Cream": 80,
    "Sandwich": 90
}

# SIDEBAR

page = st.sidebar.radio(
    "Navigation",
    [
        "Order Food",
        "Customer Analytics",
        "Demand Prediction",
        "AI Recommendations",
        "AI Chatbot Waiter"
    ]
)

# ORDER FOOD PAGE


if page == "Order Food":

    st.header("🛒 Order Food")

    customer = st.text_input("Customer Name")

    food_item = st.selectbox(
        "Select Food",
        list(menu.keys())
    )

    quantity = st.number_input(
        "Quantity",
        min_value=1,
        step=1
    )

    total_price = menu[food_item] * quantity

    st.write(f"### Total: ₹{total_price}")

    if st.button("Place Order"):

        order_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        cursor.execute("""
        INSERT INTO orders
        (customer, item, quantity, price, order_time)
        VALUES (?, ?, ?, ?, ?)
        """, (
            customer,
            food_item,
            quantity,
            total_price,
            order_time
        ))

        conn.commit()

        st.success("✅ Order Placed Successfully!")


# CUSTOMER ANALYTICS


elif page == "Customer Analytics":

    st.header("📊 Customer Analytics")

    df = pd.read_sql_query(
        "SELECT * FROM orders",
        conn
    )

    if len(df) > 0:

        st.subheader("All Orders")
        st.dataframe(df)

        # Revenue Chart
        revenue = df.groupby("item")["price"].sum().reset_index()

        fig = px.bar(
            revenue,
            x="item",
            y="price",
            title="Revenue by Food Item"
        )

        st.plotly_chart(fig)

        # Most ordered item
        popular = df.groupby("item")["quantity"].sum().reset_index()

        fig2 = px.pie(
            popular,
            names="item",
            values="quantity",
            title="Most Ordered Food"
        )

        st.plotly_chart(fig2)

    else:
        st.warning("No data available!")

# DEMAND PREDICTION

elif page == "Demand Prediction":

    st.header("📈 Food Demand Prediction")

    df = pd.read_sql_query(
        "SELECT * FROM orders",
        conn
    )

    if len(df) > 5:

        item_sales = df.groupby("item")["quantity"].sum().reset_index()

        item_sales["day"] = np.arange(len(item_sales))

        X = item_sales[["day"]]
        y = item_sales["quantity"]

        model = LinearRegression()
        model.fit(X, y)

        future_day = [[len(item_sales) + 1]]

        prediction = model.predict(future_day)

        st.write("### Predicted Future Demand")

        st.metric(
            "Expected Orders",
            int(prediction[0])
        )

        fig = px.line(
            item_sales,
            x="day",
            y="quantity",
            title="Demand Trend"
        )

        st.plotly_chart(fig)

    else:
        st.warning("Not enough data for prediction!")

# FOOD RECOMMENDATION SYSTEM


elif page == "AI Recommendations":

    st.header("🤖 Food Recommendation System")

    mood = st.selectbox(
        "Select Your Mood",
        [
            "Happy",
            "Hungry",
            "Party",
            "Light Food",
            "Cold Weather"
        ]
    )

    recommendations = {
        "Happy": ["Pizza", "Burger"],
        "Hungry": ["Pasta", "Pizza"],
        "Party": ["French Fries", "Coke"],
        "Light Food": ["Sandwich"],
        "Cold Weather": ["Pasta", "Burger"]
    }

    st.success(
        f"Recommended Foods: {', '.join(recommendations[mood])}"
    )

# AI CHATBOT WAITER

elif page == "AI Chatbot Waiter":

    st.header("🤖 AI Chatbot Waiter")

    user_input = st.text_input(
        "Ask the waiter anything..."
    )

    if user_input:

        msg = user_input.lower()

        if "recommend" in msg:
            st.success(
                "🍕 I recommend Pizza and Pasta today!"
            )

        elif "cheap" in msg:
            st.success(
                "🍟 French Fries and Coke are budget-friendly!"
            )

        elif "healthy" in msg:
            st.success(
                "🥪 Sandwich is a lighter option!"
            )

        elif "spicy" in msg:
            st.success(
                "🌶️ Pasta is our spicy special!"
            )

        else:
            st.info(
                "😊 Welcome! Ask about recommendations, healthy food, spicy food, or budget meals."
            )

'''

with open("app.py", "w") as f:
    f.write(app_code)

print("AI Food App Created!")
```
